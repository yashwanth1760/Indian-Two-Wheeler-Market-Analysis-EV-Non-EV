# Best Bikes in India (bikewala Website): Web Scraping and Data Analysis Project

## Objective

1. **Feature-Wise Bike Analysis**
   - Analyze bikes available in India based on their price, engine capacity (CC), mileage, maximum power, torque, fuel efficiency, and other specifications.
   - Compare bikes based on performance, comfort, safety features, fuel economy, and maintenance-related factors.
   - Identify bikes that offer the best combination of features, performance, and value for money.

2. **Brand-Wise and Model-Wise Bike Analysis**
   - Compare popular bike brands and models in India based on their prices, specifications, mileage, and features.
   - Analyze differences between commuter bikes, sports bikes, cruisers, adventure bikes, and touring bikes.
   - Identify the best-performing brands and models within different price ranges and riding categories.

3. **Price-Wise Bike Analysis**
   - Categorize bikes into different price segments and compare their features and specifications.
   - Identify the best bikes under ₹1 lakh, ₹1.5 lakh, ₹2 lakh, and ₹3 lakh.
   - Analyze the relationship between bike price, engine capacity, mileage, power, and available features.

4. **Best Bike Recommendation**
   - Rank bikes based on multiple criteria, including price, mileage, performance, safety, comfort, and features.
   - Develop a scoring system to identify the best bikes for different requirements, such as daily commuting, long-distance touring, sports riding, and fuel efficiency.
   - Recommend the best overall bikes and the best value-for-money options in each category using data-driven analysis.

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd, numpy as np
import re
import time
```

```python
models_nonev = []
ratings_noev = []
members_noev = []
ccs_noev = []
kmpls_noev = []
bhps_noev = []
kgs_noev = []
prices_noev = []
```

```python
models_ev = []
ratings_ev = []
members_ev = []
kmphs_ev = []
kms_ev = []
kgs_ev = []
kwhs_ev = []
prices_ev = []
```

#### Best Scooters Extraction

```python
for page_no in range(1,16):

    url = f"https://www.bikewale.com/best-scooters-in-india/?sortOrder1=desc&sortField1=modelPopularity&pageNumber={page_no}"
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/137.0.0.0 Safari/537.36",
        "Accept-Language": "en-US,en;q=0.9",
        "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/webp,/;q=0.8",
        "Referer": "https://www.google.com/"
    }
    time.sleep(3)
    page = requests.get(url, headers=headers)
    soup = BeautifulSoup(page.text,"html.parser")
    cards_list = soup.find_all("div",class_="o-f o-hl o-aE")

    for card in cards_list:

        ev = card.select_one("div.o-C.o-d.o-k.o-f5.o-ew.uZURE8.o-gS.o-ib+span")

        if ev:
            
            brand_model = card.select_one("span.o-jJ.o-j3.o-js.o-ez") #1
            
            if brand_model:
                models_ev.append(brand_model.text)
            else:
                models_ev.append(np.nan)
                

            rating_no = card.select_one("div.o-eg.o-f.o-aE")  #2
            
            if rating_no:
                text = rating_no.text
                rating_no_reg = re.findall(r"(\d+\.\d+)/5",text)
                rating_mem_reg = re.findall(r"/5(\d+)",text)

                if rating_no_reg:
                    ratings_ev.append(float(rating_no_reg[0]))
                else:
                    ratings_ev.append(np.nan)
                
                if rating_mem_reg:
                    members_ev.append(int(rating_mem_reg[0]))
                else:
                    members_ev.append(np.nan)
            else:
                ratings_ev.append(np.nan)
                members_ev.append(np.nan)


            price = card.select_one("span.o-jJ.o-jr.o-j3.o-kJ")   #3

            if price:
                
                text = price.text
                price_reg = text.split()[1].replace(",","")
                prices_ev.append(price_reg)
                
            else:
                prices_ev.append(np.nan)

                

            list_specs = card.select_one("div.o-o.FTJfQA.o-iT.o-ei")  #4

            kmph = np.nan
            km = np.nan
            kg = np.nan
            kwh = np.nan
            
            if list_specs:
                specs = list_specs.get_text(strip=True).split("|")
            
                for spec in specs:
                    spec = spec.strip().lower()
            
                    if "kmph" in spec:
                        kmph = spec
            
                    elif "kwh" in spec:
                        kwh = spec
            
                    elif "km" in spec:
                        km = spec
            
                    elif "kg" in spec:
                        kg = spec
            
            kmphs_ev.append(kmph)
            kms_ev.append(km)
            kgs_ev.append(kg)
            kwhs_ev.append(kwh)

        
                        
        else:
            
            brand_model = card.select_one("span.o-jJ.o-j3.o-js.o-ez") #1
            
            if brand_model:
                models_nonev.append(brand_model.text)
            else:
                models_nonev.append(np.nan)


            rating_no = card.select_one("div.o-eg.o-f.o-aE")  #2
            
            if rating_no:
                text = rating_no.text
                rating_no_reg = re.findall(r"(\d+\.\d+)/5",text)
                rating_mem_reg = re.findall(r"/5(\d+)",text)

                if rating_no_reg:
                    ratings_noev.append(float(rating_no_reg[0]))
                else:
                    ratings_noev.append(np.nan)

                if rating_mem_reg:
                    members_noev.append(int(rating_mem_reg[0]))
                else:
                    members_noev.append(np.nan)
            else:
                ratings_noev.append(np.nan)
                members_noev.append(np.nan)
                

            price = card.select_one("span.o-jJ.o-jr.o-j3.o-kJ")   #3

            if price:
                
                text = price.text
                price_reg = text.split()[1].replace(",","")
                prices_noev.append(price_reg)
                
            else:
                prices_noev.append(np.nan)


            list_specs = card.select_one("div.o-o.FTJfQA.o-iT.o-ei")  #4
            
            cc = np.nan
            kmpl = np.nan
            bhp = np.nan
            kg = np.nan
            
            if list_specs:
                specs = list_specs.get_text(strip=True).split("|")
            
                for spec in specs:
                    spec = spec.strip().lower()
            
                    if "cc" in spec:
                        cc = spec
                    elif "kmpl" in spec:
                        kmpl = spec
                    elif "bhp" in spec:
                        bhp = spec
                    elif "kg" in spec:
                        kg = spec
            
            ccs_noev.append(cc)  
            kmpls_noev.append(kmpl)
            bhps_noev.append(bhp)
            kgs_noev.append(kg)
            

            
```
