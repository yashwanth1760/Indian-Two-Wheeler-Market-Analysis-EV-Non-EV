# Best Bikes in India (bikewala Website): Web Scraping and Data Analysis Project

## Objective

1. **Feature-Wise Bike Analysis**
   - Analyze bikes available in India based on their price, engine capacity (CC), mileage, maximum power, torque, fuel efficiency, and other specifications.
   - Compare bikes based on performance, comfort, safety features, fuel economy, and maintenance-related factors.
   - Identify bikes that offer the best combination of features, performance, and value for money.

2. **Brand-Wise and Model-Wise Bike Analysis**
   - Compare popular bike brands and models in India based on their prices, specifications, mileage, and features.
   - Analyze differences between commuter bikes, sports bikes, cruisers, adventure bikes, and touring bikes.
   - Identify the best-performing brands and models within different price ranges and riding categories.

3. **Price-Wise Bike Analysis**
   - Categorize bikes into different price segments and compare their features and specifications.
   - Identify the best bikes under ₹1 lakh, ₹1.5 lakh, ₹2 lakh, and ₹3 lakh.
   - Analyze the relationship between bike price, engine capacity, mileage, power, and available features.

4. **Best Bike Recommendation**
   - Rank bikes based on multiple criteria, including price, mileage, performance, safety, comfort, and features.
   - Develop a scoring system to identify the best bikes for different requirements, such as daily commuting, long-distance touring, sports riding, and fuel efficiency.
   - Recommend the best overall bikes and the best value-for-money options in each category using data-driven analysis.

```python
df["Engine_CC"] = pd.to_numeric(df["Engine_CC"].str.replace("cc","",case = False, regex=True),errors = "coerce")
df["Mileage"] = pd.to_numeric(df["Mileage"].str.replace("kmpl","",case = False, regex=True),errors = "coerce")
df["Power"] = pd.to_numeric(df["Power"].str.replace("bhp","",case = False, regex=True),errors = "coerce")
df["Weight"] = pd.to_numeric(df["Weight"].str.replace("kg","",case = False, regex=True),errors = "coerce")
```

```python
df.dtypes
```

```python
multi_word_brands = ["Royal Enfield", "Moto Morini", "Moto Guzzi", "QJ Motor", "Brixton Motorcycles"]

def get_brand(name):
    name = name.strip()
    for b in multi_word_brands:
        if name.startswith(b):
            return b
    return name.split()[0]

df["Brand"] = df["Model"].apply(get_brand)
```

```python
df = df.rename(columns = {"Price": "Price(Rs)", "Mileage": "Mileage(kmpl)","Engine_CC":"Engine(cc)","Power": "Power(bhp)" ,"Weight":"Weight(kg)"})
```

```python
df = df[["Brand" ,"Model" ,"Rating"	,"Rating_Members" ,"Price(Rs)",	"Engine(cc)" , "Mileage(kmpl)" ,"Power(bhp)" ,"Weight(kg)"]]
```

```python
df[df.duplicated(keep= False)].sort_values(by="Model")
```

```python
df.drop_duplicates(inplace = True)
```

```python
df[df[["Engine(cc)","Mileage(kmpl)"	,"Power(bhp)",	"Weight(kg)"]].isna().sum(axis=1) >= 3 ]
```

```python
df = df[~(df[["Engine(cc)","Mileage(kmpl)","Power(bhp)",	"Weight(kg)"]].isna().sum(axis=1) >= 3) ]
```

```python
df.shape
```

```python
df.isna().sum()
```

```python
df.isna().mean()*100
```

```python
df[df[["Rating_Members","Rating"]].isna().sum(axis=1)>=2].count()
```

```python
df.info()
```

```python
# Step 1: Create engine groups
df["Engine_group"] = pd.cut(
    df["Engine(cc)"],
    bins=[0, 125, 160, 250, 400, 700, 1000, 1300, 3000],
    labels=[
        "upto125", "126-160", "161-250", "251-400",
        "401-700", "701-1000", "1001-1300", "1300plus"
    ]
)
```

```python
def fill_missing(df, column, tries):
    for group in tries:
        middle = df.groupby(group, observed=True)[column].transform("median")
        df[column] = df[column].fillna(middle)


fill_missing(df, "Power(bhp)", [["Brand", "Engine_group"], ["Engine_group"]])
fill_missing(df, "Weight(kg)", [["Brand", "Engine_group"], ["Engine_group"]])
fill_missing(df, "Mileage(kmpl)", [["Brand", "Engine_group"], ["Engine_group"]])
```

```python
df["Rating_Imputed"] = df["Rating"].isna()

df["Rating"] = df["Rating"].fillna(
    df.groupby("Brand")["Rating"].transform("median")
)

df["Rating"] = df["Rating"].fillna(df["Rating"].median())

df["Rating_Clean"] = df["Rating"].where(~df["Rating_Imputed"])
```

```python
df["Rating_Members"] = df["Rating_Members"].fillna(0)
```

```python
df[["Rating_Members","Rating","Rating_Imputed","Rating_Clean"]]
```

```python
print((df.isnull().mean() * 100).round(2))
```

```python
df.drop(columns=["Engine_group"], inplace=True)
```
